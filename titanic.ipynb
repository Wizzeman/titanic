```python
# remove the 'PassengerId', 'Name' and Ticket columns
df = df.drop(columns=['PassengerId','Name', 'Ticket'])
```

```python
# get the the unige values from 'Sex' column
df['Sex'].unique()
```

```python
# get the the unige values from 'Cabin' column
df["Cabin"].unique()
```

```python
# get the the unige values from 'Embarked' column
df['Embarked'].unique()
```

```python
# converting the categorical values to numerical values (0, 1)
binary_map = {'male' : 1, 'female' : 0}
df['Sex'] = df['Sex'].map(binary_map)

df
```

```python
# converted categorical variables into dummy variables
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = pd.get_dummies(df, columns=['Embarked'], dtype=int)
df.head()
```

```python
# extract the cabin (first letter) from the Cabin column
df['Cabin'] = df['Cabin'].str[0]

# replace missing cabin values with 'Unknown'
df['Cabin'] = df['Cabin'].fillna('Unknown')

# converted categorical variables into dummy variables
df = pd.get_dummies(df, columns=['Cabin'], dtype=int)
df.head()
```
